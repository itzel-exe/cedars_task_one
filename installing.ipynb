## Installation and Setup

Before running the analysis, make sure R and the required packages are installed.

1. Download or clone this repository.
2. Open `analysis.R` in RStudio.
3. Update the `data_dir` path in the script to the location where you saved the project files:

```r
data_dir <- "C:/Users/YourName/Path/To/assignment_one"
```

For example:

```r
data_dir <- "C:/Users/Itzel/Desktop/assignment_one"
```

4. Make sure the following files are located in that folder:

   * `master_data.csv`
   * `master_data_variables.xlsx`
   * `t05xc.csv`
   * `t016.csv`
   * `analysis_functions.R`

5. Run `analysis.R` in RStudio.

**Important:** Replace the `data_dir` path with the path on your own computer. The analysis code uses this path to locate the input datasets and supporting files.
